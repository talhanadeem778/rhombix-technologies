```python
import pandas as pd
dataset1 = pd.read_csv(r"C:\Users\Admin\Documents\datasets\sales_data.csv")
print(dataset1)
```

```python

dataset2 = pd.read_excel(r"C:\Users\Admin\Documents\datasets\Product-Sales-Region.csv")
print(dataset2)
```

```python
merge_dataset = pd.merge(dataset1 , dataset2 , on = 'Region')
print(merge_dataset)
```

```python
revenue = merge_dataset['Sales_Amount'] * merge_dataset['Quantity_Sold']
total_revenue = merge_dataset['revenue'].sum()
print("total revenue will be" , total_revenue)
```

```python

profit_margin = merge_dataset['Sales_Amount'] /  merge_dataset['TotalPrice'].sum()
print("total margin will be" , profit_margin)
```

```python
#total profit
total_profit = merge_dataset['Sales_Amount'] -  merge_dataset['TotalPrice'].sum()
print("total profil will be" , total_profit)
```

```python
#total sales
total_sales = merge_dataset['Sales_Amount'].sum()
print("total sales is" , total_sales) 
```

```python
#avg sales
avg_sales = merge_dataset['Sales_Amount'].mean()
print("total sales is" , avg_sales)
```

```python
#total quanntity sold 
total_quantity_sold = merge_dataset['Quantity_Sold'].sum()
print("total quantity_sold is " , total_quantity_sold)
```

```python
#total orders 
total_orders = dataset2['OrderID'].count()
print("total orders will be " , total_orders)
```

```python
#revenue by region 
revenue_by_region = merge_dataset.groupby('Region')['revenue'].sum()
print(revenue_by_region)
```

```python
#revenue by product
revenue_by_product = merge_dataset.groupby('Product_Category')['revenue'].sum()
print(revenue_by_product)
```

```python
#top selling products 
top_selling_products = merge_dataset.groupby('Product_Category')['Quantity_Sold'].sum().sort_values().head(10)
print(top_selling_products)
```
