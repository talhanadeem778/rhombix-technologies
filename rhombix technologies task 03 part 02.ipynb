```python
import pandas as pd
dataset1 = pd.read_csv(r"C:\Users\Admin\Documents\datasets\sales_data.csv")
print(dataset1)
```

```python
dataset2 = pd.read_excel(r"C:\Users\Admin\Documents\datasets\Product-Sales-Region.csv")
print(dataset2)
```

```python
df3 = pd.merge(dataset1 , dataset2 , on = "Region")
print(df3)
```

```python
#z_test
from scipy.stats import zscore , norm 
pop_mean = 1223.45
pop_std = 3435.44

mean = df3['TotalPrice'].mean()
size = len(df3)
z_score =  (mean - pop_mean) / (pop_std - size**0.5)
p_value = 2 * norm.sf(abs(z_score))

print("z_score is" , z_score)
print("p_value is " , p_value)

```

```python
#f_test

import scipy.stats as stats
var1 = df3['Unit_Cost']
var2 = df3['Unit_Price']

f_test = var1 / var2

dof1 = len(df3["Unit_Cost"]) - 1
dof2 = len(df3['Unit_Price']) - 1

p_value = 1 - stats.f.cdf(f_test, dfn=dof1, dfd=dof2)

print("f_test is " , f_test)
print("dof1 is " , dof1)
print("dof2 is " , dof2)
print("p value is " , p_value)

```

```python
#t_test 
import scipy.stats as stats
g1 = df3[df3['Unit_Cost'] == '152.75']['TotalPrice']
g2 = df3[df3['Unit_Cost'] == '2632.58']['TotalPrice']
t_test, p_value = stats.ttest_ind(g1 , g2)
print("t_test is " , t_test)
print("p_value is " , p_value)

before = df3[df3['Unit_Cost'] == '152.75']['TotalPrice']
after = df3[df3['Unit_Cost'] == '2632.58']['TotalPrice']
t_stat, p_value = stats.ttest_rel(before, after)

print("Paired t-test t-statistic:", t_stat)
print("Paired t-test p-value:", p_value)
```

```python
from scipy.stats import chi2_contingency

# Build a contingency table from two categorical variables.
data = pd.crosstab(df3['Region'], df3['Sales_Channel'])
chi2, p, dof, expected = chi2_contingency(data)

print("Observed table:")
print(data)
print("chi2 is", chi2)
print("p is", p)
print("dof is", dof)
print("expected frequency is", expected)

if p < 0.05:
    print("good hephothesis")
else:
    print("bad hypothesis")
```

```python
# ANOVA test
import statsmodels.api as sm
from statsmodels.formula.api import ols

model = ols('TotalPrice ~ C(Region)', data=df3).fit()
print(model.summary())
anova_table = sm.stats.anova_lm(model, typ=2)
print(anova_table)
```
